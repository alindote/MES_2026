```python
import numpy as np
import numpy.random as rnd
import matplotlib.pyplot as plt

def f(x):
    return 2+x/4+np.sin(2*x)

N = 10**5 # número de pontos a amostrar

# gerar um array de aleatórios de tamanho N
x = 2*np.pi*rnd.random(N)  # o eixo xx varia entre 0 e 2pi

# criar array com o valor da função nos pontos gerados
y = f(x)

# listar os pontos aleatórios e os valores da função
# usar só para N pequeno
#print("x = ", x)
#print("y = ", y)

# calcular a estimativa de MC da função
i_mc = (2*np.pi)*y.sum()/N
i_calc = 17.50

print('Estimativa de MC para o integral:', i_mc)
print('Valor exacto do integral:', i_calc)
print('Erro relativo:', abs(i_mc-i_calc)/i_calc*100,'%')

# se quisermos representar a função
plt.figure(1)
x = np.arange(0.01, 2*np.pi, 0.01)
plt.plot(x, f(x), 'k', linewidth=2)
plt.show()
```
