```python

import numpy as np
import matplotlib.pyplot as plt

# Read the output file, ignoring the two comment lines
data = np.loadtxt("eDep_scorer.out", delimiter=",", comments="#")

# Extract the columns
x = data[:, 0].astype(int)
y = data[:, 1].astype(int)
z = data[:, 2].astype(int)
E = data[:, 3]       # Total deposited energy (MeV)
E2 = data[:, 4]      # Accumulated squared values
entries = data[:, 5] # Number of entries

print(f"Number of voxels: {len(E)}")
print(f"Total deposited energy: {np.sum(E):.3f} MeV")

# Geometry
N = 45
centre = 22
```

```python
# Energy spectrum: deposited energy per voxel
plt.figure(figsize=(8, 6))

plt.hist(E, bins=100, color="steelblue", histtype="stepfilled")

plt.xlabel("Deposited energy per voxel (MeV)")
plt.ylabel("Number of voxels")
plt.title("Energy deposition spectrum")
plt.tight_layout()
plt.show()
```

## Other interesting plots

```python
# --------------------------------------------------
# 1. XY projection: sum energy over all z layers
# --------------------------------------------------

xy_energy = np.zeros((N, N))
np.add.at(xy_energy, (y, x), E)

plt.figure(figsize=(8, 7))
plt.imshow(
    xy_energy,
    origin="lower",
    extent=(-0.5, 44.5, -0.5, 44.5),
    aspect="equal",
    cmap="inferno"
)
plt.colorbar(label="Deposited energy (MeV)")
plt.xlabel("Voxel x")
plt.ylabel("Voxel y")
plt.title("XY projection: energy summed over z")
plt.tight_layout()
plt.show()
```

```python
# --------------------------------------------------
# 2. Deposited energy versus radial distance
# --------------------------------------------------

R = np.sqrt((x - centre)**2 + (y - centre)**2)

plt.figure(figsize=(8, 6))
plt.scatter(R, E, s=5, alpha=0.5)
plt.xlabel("Radius from beam axis (voxels)")
plt.ylabel("Deposited energy per voxel (MeV)")
plt.title("Energy deposition versus radius")
plt.tight_layout()
plt.show()
```

```python
# --------------------------------------------------
# 3. Deposited energy versus depth z
# --------------------------------------------------

plt.figure(figsize=(8, 6))
plt.scatter(z, E, s=5, alpha=0.5)
plt.xlabel("Voxel z")
plt.ylabel("Deposited energy per voxel (MeV)")
plt.title("Energy deposition versus depth")
plt.tight_layout()
plt.show()
```
